## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1 · 조건문 — 값에 따라 실행할 길을 고른다</mark>

로그 한 건을 보고 정상인지 확인이 필요한지 판단합니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.1 실패 로그인지 확인하고 싶을 때</mark>

조건이 참일 때만 들여쓴 줄이 실행됩니다.


```python
# 로그인 로그 한 건을 준비합니다.
log = {"user": "kim01", "result": "FAIL", "fail_count": 2}

if log["result"] == "FAIL":  # 실패 로그인지 확인
    print("실패 로그입니다")  # 조건이 참일 때 실행
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.2 두 갈래로 나누고 싶을 때</mark>

`if`가 거짓이면 `else` 쪽이 실행됩니다.


```python
# 정상 로그인 로그 한 건을 준비합니다.
log = {"user": "lee02", "result": "OK", "fail_count": 0}

if log["result"] == "FAIL":  # 실패인지 확인
    print("확인 필요")  # 실패일 때 실행
else:  # 실패가 아닐 때 실행
    print("정상")
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.3 여러 단계로 나누고 싶을 때</mark>

위에서부터 검사하고 처음 참인 한 갈래만 실행됩니다.


```python
fail_count = 5  # 현재 실패 횟수

if fail_count >= 5:  # 5회 이상인지 확인
    print("높음")
elif fail_count >= 2:  # 2회 이상인지 확인
    print("주의")
else:  # 두 조건에 모두 해당하지 않을 때 실행
    print("낮음")
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.4 조건을 함께 확인하고 싶을 때</mark>

`and`는 두 조건이 모두 참이어야 합니다. `or`는 하나만 참이어도 됩니다.


```python
# 로그인 로그 한 건을 준비합니다.
log = {"user": "park03", "action": "LOGIN", "result": "FAIL", "fail_count": 4}

if log["action"] == "LOGIN" and log["result"] == "FAIL":  # 두 조건을 모두 확인
    print("로그인 실패")

if log["fail_count"] >= 5 or log["user"] == "admin":  # 둘 중 하나를 확인
    print("즉시 확인")
else:
    print("계속 관찰")
```

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2 · 반복문 — 같은 일을 모든 값에 적용한다</mark>

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.1 리스트를 처음부터 끝까지 보고 싶을 때</mark>

```python
users = ["kim01", "lee02", "park03"]  # 확인할 계정 목록

for user in users:  # 계정을 하나씩 꺼내 반복
    print(user)
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.2 로그 딕셔너리를 하나씩 보고 싶을 때</mark>

한 번 돌 때마다 `log`에는 딕셔너리 하나가 들어갑니다.


```python
# 확인할 로그인 로그 세 건을 준비합니다.
logs = [
    {"user": "kim01", "result": "FAIL"},
    {"user": "lee02", "result": "OK"},
    {"user": "park03", "result": "FAIL"},
]

for log in logs:  # 로그를 하나씩 꺼내 반복
    print(log["user"], log["result"])
```

<table width="100%" bgcolor="#A5D6A7"><tr><td bgcolor="#A5D6A7"><h3><font color="#1B5E20">✍️ 문제 2-2 · 실패 결과만 출력하기</font></h3></td></tr></table>

| | |
|---|---|
| **주어지는 값** | 로그 딕셔너리 3건 |
| **🎯 나와야 하는 결과** | `kim01`<br>`park03` |

<details>
<summary>▸ 힌트 보기</summary>

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.3 실패한 계정만 모으고 싶을 때</mark>

빈 리스트를 만들고, 조건을 통과한 값만 `append()`합니다.


```python
# 확인할 로그인 로그 세 건을 준비합니다.
logs = [
    {"user": "kim01", "result": "FAIL"},
    {"user": "lee02", "result": "OK"},
    {"user": "park03", "result": "FAIL"},
]

failed_users = []  # 실패 계정을 담을 빈 목록
for log in logs:  # 로그를 하나씩 꺼내 반복
    if log["result"] == "FAIL":  # 실패 로그만 선택
        failed_users.append(log["user"])  # 계정을 목록에 추가

print(failed_users)
print(f"실패 {len(failed_users)}건")  # 목록 길이로 실패 건수 확인
```

| | |
|---|---|
| **주어지는 값** | 로그 딕셔너리 3건 |
| **🎯 나와야 하는 결과** | `['kim01', 'park03']` |

<details>
<summary>▸ 힌트 보기</summary>

빈 리스트는 반복문 위에 만들고, 조건 안에서 `append()`합니다.

</details>


```python
logs = [
    {"user": "kim01", "result": "FAIL"},
    {"user": "lee02", "result": "OK"},
    {"user": "park03", "result": "FAIL"},
]
failed_users = []
for log in logs:
  if log["result"] == "FAIL":
    failed_users.append(log["user"])
print(failed_users)
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.5 조건이 참인 동안 되풀이하고 싶을 때</mark>

`while`은 조건이 참인 동안 반복합니다. 조건에 쓰는 값이 바뀌어야 끝납니다.


```python
fail_count = 0  # 반복을 시작할 값

while fail_count < 3:  # 3보다 작은 동안 반복
    fail_count = fail_count + 1  # 1씩 증가
    print(fail_count)
```

<details>

</details>


```python
count = 0

while count < 3:
  count = count + 1
  print(count)
```

실패한 사람이 몇 명인지, 각자 몇 번 실패했는지를 구분합니다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.1 중복을 지우고 싶을 때</mark>

`set()`은 같은 값을 하나만 남깁니다. 결과의 순서는 달라질 수 있습니다.


```python
# 중복이 있는 실패 계정 목록을 준비합니다.
failed_users = ["kim01", "kim01", "lee02", "park03", "kim01"]
unique_users = set(failed_users)  # 중복 제거

print(unique_users)
print(f"실패 계정 {len(unique_users)}명")  # 서로 다른 계정 수 확인
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.2 값마다 횟수를 세고 싶을 때</mark>

`Counter`는 값마다 나온 횟수를 딕셔너리처럼 돌려줍니다.


```python
from collections import Counter  # 항목별 횟수를 세는 도구

# 중복이 있는 실패 계정 목록을 준비합니다.
failed_users = ["kim01", "kim01", "lee02", "park03", "kim01"]
counts = Counter(failed_users)  # 계정별 횟수 계산

print(counts)
print(counts["kim01"])  # kim01의 횟수만 확인
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.3 계정과 횟수를 함께 돌고 싶을 때</mark>

`.items()`는 키와 값을 한 쌍씩 꺼냅니다.


```python
from collections import Counter  # 항목별 횟수를 세는 도구

# 실패 계정별 횟수를 계산합니다.
counts = Counter(["kim01", "kim01", "lee02", "park03", "kim01"])
for user, count in counts.items():  # 계정과 횟수를 함께 반복
    print(f"{user}: {count}회")
```

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.4 기준 이상인 계정만 찾고 싶을 때</mark>

기준값에는 대문자 이름을 붙여 한 곳에서 바꿉니다.


```python
from collections import Counter  # 항목별 횟수를 세는 도구

THRESHOLD = 2  # 확인이 필요한 기준 횟수
# 실패 계정별 횟수를 계산합니다.
counts = Counter(["kim01", "kim01", "lee02", "park03", "kim01"])

for user, count in counts.items():  # 계정과 횟수를 함께 반복
    if count >= THRESHOLD:  # 기준 이상인지 확인
        print(f"확인 필요: {user} · {count}회")
```

<table width="100%" bgcolor="#A5D6A7"><tr><td bgcolor="#A5D6A7"><h3><font color="#1B5E20">✍️ 문제 3-5 · 기준 이상만 출력하기</font></h3></td></tr></table>

실패가 2회 이상인 계정만 출력하시오.

| | |
|---|---|
| **주어지는 값** | `THRESHOLD = 2` · `counts` |
| **🎯 나와야 하는 결과** | `kim01` |

<details>
<summary>▸ 힌트 보기</summary>

반복문 안에서 `count >= THRESHOLD`를 확인합니다.

</details>


```python
from collections import Counter
THRESHOLD = 2
counts = Counter(["kim01", "kim01", "lee02"])

for user, count in counts.items():
  if count >= THRESHOLD:
    print(user)
```

<table width="100%" bgcolor="#A5D6A7"><tr><td bgcolor="#A5D6A7"><h3><font color="#1B5E20">✍️ 문제 3-6 · 기준 바꿔 다시 보기</font></h3></td></tr></table>

기준을 3으로 바꾸고, 3회 이상인 계정과 횟수를 출력하시오.

| | |
|---|---|
| **주어지는 값** | 계정별 실패 횟수 |
| **🎯 나와야 하는 결과** | `park03 · 3회` |

<details>
<summary>▸ 힌트 보기</summary>

기준값은 위 한 줄에서만 바꾸고 아래 코드는 그대로 둡니다.

</details>


```python
from collections import Counter
THRESHOLD = 3
counts = Counter(["kim01", "kim01", "lee02", "park03", "park03", "park03"])

for user, count in counts.items():
  if count >= THRESHOLD:
    print(f"{user}, {count}회")
```
