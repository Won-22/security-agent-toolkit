## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1 · 변수 — 값에 이름을 지정한다</mark>

변수(variable)는 **값 하나에 붙여 둔 이름**이다. 값을 계속 손에 들고 있을 수 없으니 이름을 지어 두는 것이다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.1 값에 이름을 지정하고 싶을 때</mark>

이름을 왼쪽에, 값을 오른쪽에 씁니다.

````python
# 이 셀의 목적 — 값 하나에 이름을 지정하고 꺼내 본다
# 글자는 반드시 따옴표로 감쌉니다.

user = "kim01"        # 따옴표 = 글자라는 표시
print(user)           # 이름을 부르면 그 값이 나온다
````

> **🤔 이런 의문이 들 수 있습니다**
>
> `=` 는 「같다」 아닌가? 라는 의문이 들 수 있다.
>
> 수학의 등호와 다르다. 파이썬의 `=` 는 **「오른쪽 값을 왼쪽 이름에 넣어라」** 는 명령이다.
> 방향이 있다 — **오른쪽에서 왼쪽으로.** 그래서 `3 = user` 라고 쓰면 오류가 난다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.2 값을 바꾸고 싶을 때</mark>

같은 이름에 새 값을 넣으면 앞엣것은 사라집니다.

````python
# 이 셀의 목적 — 같은 이름에 새 값을 넣으면 어떻게 되는지 본다
# 지금 값 3 을 꺼내 1 을 더하고 같은 이름에 다시 넣습니다.

fail_count = 3                       # 로그인 실패 3회
fail_count = fail_count + 1          # 한 번 더 실패 → 4 가 되어 다시 들어간다
print(fail_count)                    # 4
````

> **⚠ 이렇게 쓰면 안 됩니다**
>
> ```python
> user = kim01
> ```
> ```
> NameError: name 'kim01' is not defined
> ```
> 따옴표를 빼먹으면 파이썬은 `kim01` 을 **또 다른 변수 이름**으로 읽는다.
> 그런 이름이 없으니 `NameError` 가 난다. 오늘 제일 많이 볼 오류다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.3 로그 한 줄을 변수로 나누고 싶을 때</mark>

보안 로그 한 줄이 변수 셋이 됩니다.

````python
# 이 셀의 목적 — 보안 로그 한 줄을 변수 여러 개로 쪼개 본다
# 2026-09-22 14:03:11  kim01  10.0.3.21  FAIL

user = "kim01"           # 누가
ip = "10.0.3.21"         # 어디서
result = "FAIL"          # 결과가 어땠나
print(user, ip, result)  # 쉼표로 나누면 한 줄에 나란히 찍힌다
````

### <mark style="display:block; background:#bbdefb; color:#1a1a1a; padding:6px 12px; border-radius:4px">1.정리 📋 한눈에</mark>

| 쓰는 법 | 뜻 |
|---|---|
| `user = "kim01"` | user 라는 이름에 글자 kim01 을 넣는다 |
| `fail_count = 3` | 숫자는 따옴표 없이 |
| `a = a + 1` | 지금 값을 꺼내 1 더하고 다시 넣는다 |
| `print(a, b)` | 쉼표로 나누면 한 줄에 나란히 |

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2 · 자료형 — 값의 갈래</mark>

자료형(type)은 **그 값이 숫자인지 글자인지 하는 갈래**다. 보안 로그에서 네 가지를 다 만난다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.1 값의 갈래를 만들어 보고 싶을 때</mark>

네 갈래를 하나씩 만들어 찍어 봅니다.

````python
# 이 셀의 목적 — 네 가지 자료형을 하나씩 만들어 본다
# bool 은 True / False 의 첫 글자가 대문자입니다.

fail_count = 3            # int   — 소수점 없는 수
risk_score = 0.75         # float — 소수점 있는 수
user = "kim01"            # str   — 따옴표로 감싼 글자
is_blocked = False        # bool  — True 아니면 False
print(fail_count, risk_score, user, is_blocked)
````

> **🤔 이런 의문이 들 수 있습니다**
>
> `"123"` 은 숫자인데 왜 글자지? 라는 의문이 들 수 있다.
>
> **따옴표로 둘러싸여 있으면 모두 글자(str)** 다. 안에 무엇이 들었는지는 상관없다.
> 화면에 `123` 으로 보여도 따옴표가 있으면 계산에 쓸 수 없다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.2 이 값이 무슨 갈래인지 알고 싶을 때</mark>

헷갈리면 `type()` 으로 물어보면 됩니다.

````python
# 이 셀의 목적 — 값의 자료형을 type() 으로 확인한다
# class 뒤에 나오는 것이 자료형 이름입니다.

print(type(3))            # <class 'int'>
print(type("kim01"))      # <class 'str'>
print(type(False))        # <class 'bool'>
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.3 따옴표가 있고 없고를 구분하고 싶을 때</mark>

화면에는 둘 다 7 로 보이지만 다른 값입니다.

````python
# 이 셀의 목적 — 따옴표가 자료형을 어떻게 가르는지 본다
# 아래 두 값은 화면에 똑같이 7 로 보입니다.

a = 7                     # 숫자 7
b = "7"                   # 글자 7
print(a, type(a))         # 7 <class 'int'>
print(b, type(b))         # 7 <class 'str'>
````

### <mark style="display:block; background:#bbdefb; color:#1a1a1a; padding:6px 12px; border-radius:4px">2.정리 📋 한눈에</mark>

| 자료형 | 예 | 로그에서 쓰이는 곳 |
|---|---|---|
| `int` | `3` | 실패 횟수 · 포트 번호 |
| `float` | `0.75` | 응답 시간 · 위험 점수 |
| `str` | `"kim01"` | 계정 이름 · IP · 시각 |
| `bool` | `True` · `False` | 차단 여부 · 성공 여부 |

파이썬은 자료형을 미리 적지 않는다. **넣는 값을 보고 알아서 정한다.**

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3 · f-string — 문장 안에 값을 끼워 넣는다</mark>

f-string 은 **문장 안의 특정 자리에 값을 삽입하는 방법**이다. 오늘 배우는 것 중 여덟 날 내내 제일 많이 쓴다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.1 문장 안에 값을 넣고 싶을 때</mark>

따옴표 **앞에 `f`** 를 붙이고, 값이 들어갈 자리에 **중괄호**로 이름을 적습니다.

````python
# 이 셀의 목적 — f-string 으로 문장 안에 값을 끼워 넣는다
# 따옴표 앞의 f 와 중괄호가 전부입니다.

user = "kim01"
print(f"{user}님이 로그인했습니다")
````

> **⚠ 이렇게 쓰면 안 됩니다**
>
> ```python
> user = "kim01"
> print("{user}님이 로그인했습니다")
> ```
> ```
> {user}님이 로그인했습니다
> ```
> `f` 를 빼먹으면 **중괄호가 그대로 찍힌다.** 오류가 안 나서 더 찾기 어렵다.
> 출력에 `{}` 가 보이면 `f` 를 빠뜨린 것이다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.2 값을 여러 개 넣고 싶을 때</mark>

중괄호를 여러 개 두면 됩니다. 중괄호 안에서 계산도 됩니다.

````python
# 이 셀의 목적 — 값을 여러 개 넣고 중괄호 안에서 계산까지 해 본다
# 숫자를 넣을 때도 글자로 바꿀 필요가 없습니다.

user = "kim01"
fail_count = 5
print(f"{user} 계정이 {fail_count}회 실패했습니다")
print(f"한 번 더 실패하면 차단: {fail_count + 1}회")   # 중괄호 안에서 계산
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.3 알림 문장을 만들고 싶을 때</mark>

Day 5 에 이 문장을 슬랙으로 보냅니다.

````python
# 이 셀의 목적 — 변수 네 개를 한 문장으로 조립한다
# 이 모양이 Day 5 의 알림 메시지가 됩니다.

user = "kim01"
ip = "10.0.3.21"
fail_count = 5
is_blocked = True
alert = f"[경고] {user} — {ip}에서 {fail_count}회 실패 · 차단={is_blocked}"
print(alert)
````

### <mark style="display:block; background:#bbdefb; color:#1a1a1a; padding:6px 12px; border-radius:4px">3.정리 📋 한눈에</mark>

| 쓰는 법 | 결과 |
|---|---|
| `f"{user}님"` | `kim01님` |
| `f"{a}회 · {b}건"` | 값 여러 개 |
| `f"{n + 1}회"` | 중괄호 안에서 계산 |
| `"{user}님"` | ⚠ `f` 없음 → `{user}님` 이 그대로 찍힘 |

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">4 · 형변환 — 값의 갈래를 바꾼다</mark>

형변환(casting)은 **값의 자료형을 다른 자료형으로 바꾸는 일**이다. 파일에서 읽은 값은 숫자처럼 보여도 전부 글자라서 이 작업이 꼭 필요하다.

> **🤔 이런 의문이 들 수 있습니다**
>
> 화면에 `7` 이라고 나오는데 왜 1 을 못 더하지? 라는 의문이 들 수 있다.
>
> 화면에 보이는 모양과 자료형은 다르다. `"7"` 과 `7` 은 **화면에 똑같이 7** 로 보이지만
> 앞엣것은 글자다. **글자와 숫자는 더할 수 없다.**

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">4.1 글자를 숫자로 바꾸고 싶을 때</mark>

`int()` 로 감싸면 계산할 수 있게 됩니다.

````python
# 이 셀의 목적 — 글자로 된 숫자를 int() 로 바꿔 계산한다
# int 로 감싼 순간 숫자가 됩니다.

count = int("7")          # 글자 "7" 을 숫자 7 로
print(count + 1)          # 8
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">4.2 숫자를 글자로 바꾸고 싶을 때</mark>

이어 붙이려면 글자여야 합니다.

````python
# 이 셀의 목적 — 숫자를 글자로 바꿔 이어 붙인다
# f-string 을 쓰면 이렇게 할 일이 거의 없습니다.

count = 7
print("실패 " + str(count) + "회")     # str() 로 감싸야 이어 붙는다
print(f"실패 {count}회")               # f-string 이면 그냥 된다
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">4.3 파일에서 읽은 값을 다루고 싶을 때</mark>

Day 2 에 실제로 만날 장면입니다.

````python
# 이 셀의 목적 — 파일에서 읽은 값이 글자라서 생기는 문제를 미리 본다
# split(",") 는 쉼표를 기준으로 줄을 자릅니다. Day 2 에 배웁니다.

row = "kim01,10.0.3.21,5"
parts = row.split(",")               # ['kim01', '10.0.3.21', '5']
fail_count = int(parts[2])           # "5" 를 숫자 5 로 바꾼다
if fail_count >= 5:                  # 숫자여야 크기를 비교할 수 있다
    print("차단 대상")
````

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5 · 리스트 — 여러 값을 순서대로 담는다</mark>

리스트(list)는 **값을 순서대로 줄 세워 담은 것**이다. 계정이 서른 개면 변수를 서른 개 만들 수 없으니 한 이름에 담는다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.1 여러 값을 한 이름에 담고 싶을 때</mark>

대괄호로 감싸고 쉼표로 나눕니다.

````python
# 이 셀의 목적 — 리스트를 만들고 번호로 값을 꺼낸다
# 번호는 0 부터 셉니다.

users = ["kim01", "lee02", "park03"]
print(users[0])           # kim01  ← 첫 번째가 0번
print(users[2])           # park03 ← 마지막이 2번
print(len(users))         # 3      ← len = 개수 세기
````

> **🤔 이런 의문이 들 수 있습니다**
>
> `users[2]` 가 왜 세 번째지? 라는 의문이 들 수 있다.
>
> **파이썬은 0 부터 센다.** 그래서 첫 번째가 `[0]`, 두 번째가 `[1]`, 세 번째가 `[2]` 다.
> 세 개짜리 리스트의 마지막은 `[2]` 이고 `[3]` 은 없다.
>
> 처음엔 다 헷갈린다. 며칠 쓰면 몸에 붙는다.

> **⚠ 이렇게 쓰면 안 됩니다**
>
> ```python
> users = ["kim01", "lee02", "park03"]
> print(users[3])
> ```
> ```
> IndexError: list index out of range
> ```
> 「범위를 벗어났다」는 뜻이다. 개수가 3 이면 마지막 번호는 **2** 다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.2 값을 나중에 더하고 싶을 때</mark>

빈 리스트로 시작해 하나씩 붙입니다.

````python
# 이 셀의 목적 — 빈 리스트에 값을 하나씩 더한다
# 점을 찍고 부르는 것을 메서드라고 합니다.

failed_users = []                    # 빈 대괄호 = 빈 리스트
failed_users.append("kim01")         # 뒤에 하나 붙인다
failed_users.append("lee02")
print(failed_users)                  # ['kim01', 'lee02']
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">5.3 조건에 맞는 것만 모으고 싶을 때</mark>

Day 2 에 반복문과 만나면 이 모양이 됩니다.

````python
# 이 셀의 목적 — 실패한 계정만 골라 리스트에 모으는 모양을 미리 본다
# 조건문·반복문은 Day 2 에 배웁니다. 지금은 모양만 보세요.

logs = ["kim01 FAIL", "lee02 OK", "park03 FAIL"]
failed_users = []
for line in logs:                    # 한 줄씩 꺼내서
    if "FAIL" in line:               # FAIL 이 들어 있으면
        failed_users.append(line)    # 모은다
print(failed_users)
print(f"실패 {len(failed_users)}건")
````

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">6 · 딕셔너리 — 키로 값을 꺼낸다</mark>

딕셔너리(dictionary)는 **이름(키)과 값을 짝지어 담은 것**이다. 리스트는 번호로 꺼내지만 딕셔너리는 이름으로 꺼낸다.

> **🤔 이런 의문이 들 수 있습니다**
>
> 리스트가 있는데 왜 딕셔너리가 또 필요하지? 라는 의문이 들 수 있다.
>
> 리스트는 번호로 꺼낸다. 그런데 **3번이 부서였나 등급이었나** 외우기 어렵다.
> 그래서 번호 대신 `"dept"` 같은 **이름**을 붙인다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">6.1 이름으로 값을 꺼내고 싶을 때</mark>

중괄호로 감싸고, 키와 값을 콜론으로 짝지어 둡니다.

````python
# 이 셀의 목적 — 딕셔너리를 만들고 키로 값을 꺼낸다
# 키는 이름표, 값은 그 안에 든 내용입니다.

user = {"name": "kim01", "dept": "finance", "level": "read"}
print(user["name"])       # kim01
print(user["dept"])       # finance
````

## <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7 · .get() — 없는 키를 물어도 멈추지 않는다</mark>

로그에는 **빈 칸이 늘 있다.** 어떤 줄은 IP 가 있고 어떤 줄은 없다.

없는 키를 대괄호로 꺼내면 `KeyError` 가 나면서 **거기서 프로그램이 멈춘다.** 수백 줄을 돌리다 한 줄에서 멈추면 나머지를 못 본다.

> **⚠ 이렇게 쓰면 안 됩니다**
>
> ```python
> log = {"user": "kim01"}
> print(log["ip"])
> ```
> ```
> KeyError: 'ip'
> ```
> `log` 에 `ip` 라는 키가 없어서 난다. **이 코드는 실행하지 않는다** — 아래 `.get()` 을 쓴다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.1 키가 있는지 먼저 확인하고 싶을 때</mark>

`in` 을 쓰면 그 키가 있는지 `True` · `False` 로 알려 줍니다.

````python
# 이 셀의 목적 — 딕셔너리에 그 키가 있는지 in 으로 확인한다
# 있으면 True, 없으면 False 가 나옵니다.

log = {"user": "kim01", "result": "FAIL"}    # ip 가 없다
print("user" in log)      # True
print("ip" in log)        # False
````

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.2 멈추지 않게 하고 싶을 때</mark>

`.get()` 은 없으면 `None` 을 줍니다.

````python
# 이 셀의 목적 — .get() 이 없는 키를 물어도 멈추지 않는 것을 확인한다
# 맨 아래 검사 완료가 찍히는지 보세요. 아까와 다릅니다.

log = {"user": "kim01", "result": "FAIL"}
print(log.get("ip"))                  # None = 값이 없다는 뜻
print(log.get("ip", "알수 없음"))      # 두 번째 칸 = 없을 때 대신 쓸 값
print("검사 완료")                      # 이번엔 실행된다
````

> **🤔 이런 의문이 들 수 있습니다**
>
> 그럼 항상 `.get()` 을 쓰면 되지 않나? 라는 의문이 들 수 있다.
>
> 아니다. **없으면 그게 문제인 칸**은 멈추는 게 맞다.
> 조용히 `None` 으로 넘어가면 나중에 훨씬 찾기 어려운 버그가 된다.
>
> 계정 이름은 없으면 안 되니 대괄호, IP 는 없을 수 있으니 `.get()` 이다.

### <mark style="display:block; background:#ffecb3; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.3 상황에 맞게 갈라 쓰고 싶을 때</mark>

반드시 있어야 하는 칸과 없을 수도 있는 칸을 갈라 씁니다.

````python
# 이 셀의 목적 — 대괄호와 .get() 을 상황에 맞게 갈라 쓴다
# user 는 없으면 문제다 → 대괄호. ip 는 없을 수 있다 → .get()

log = {"user": "kim01", "result": "FAIL"}
print(f"계정 {log['user']} · IP {log.get('ip', '-')} · 결과 {log['result']}")
````

### <mark style="display:block; background:#bbdefb; color:#1a1a1a; padding:6px 12px; border-radius:4px">7.정리 📋 한눈에</mark>

| 쓰는 법 | 없을 때 | 언제 쓰나 |
|---|---|---|

````python
# 이 셀의 목적 — 문제 7-1. .get() 으로 없는 키를 안전하게 꺼낸다
# 나와야 하는 결과 — 알 수 없음 / 검사 완료

log = {"user": "kim01", "result": "FAIL"}    # 주어지는 값 (그대로 두세요)

# 여기에 직접 작성 👇
print(log.get("ip", "알 수 없음"))
print("검사 완료")
````
